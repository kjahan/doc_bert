## Get documents embeddings for Jaguar

```python
# Sentences are encoded by calling model.encode()
embeddings_1 = sbert_model.encode(jaguar_animal_article)
embeddings_2 = sbert_model.encode(jaguar_car_article)
```

## Vector representation of doc using sum operation

```python
jag_anim_vec = embeddings_1[0]
for inx in range(1, len(embeddings_1)):
  jag_anim_vec += embeddings_1[inx]

jag_car_vec = embeddings_2[0]
for inx in range(1, len(embeddings_2)):
  jag_car_vec += embeddings_2[inx]


jag_anim_vec_norm = jag_anim_vec / np.linalg.norm(jag_anim_vec)
jag_car_vec_norm = jag_car_vec / np.linalg.norm(jag_car_vec)
```

## Test norms

```python
LA.norm(jag_anim_vec_norm)
```

```python
LA.norm(jag_car_vec_norm)
```

## Compute dot-product

```python
tot_sim = np.dot(jag_anim_vec_norm, jag_car_vec_norm)

print("Dot product of jaguar animal and Jaguar car vecors: {}".format(tot_sim))
```
