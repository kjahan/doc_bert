## Jaguar Cars

https://en.wikipedia.org/wiki/Jaguar_Cars

```python
p21 = "Jaguar (UK: /ˈdʒæɡjuər/, US: /ˈdʒæɡwɑːr/) is the luxury vehicle brand of Jaguar Land Rover,[1][2] a British multinational car manufacturer with its headquarters in Whitley, Coventry, England. Jaguar Cars was the company that was responsible for the production of Jaguar cars until its operations were fully merged with those of Land Rover to form Jaguar Land Rover on 1 January 2013."

p22 = "Jaguar's business was founded as the Swallow Sidecar Company in 1922, originally making motorcycle sidecars before developing bodies for passenger cars. Under the ownership of S. S. Cars Limited, the business extended to complete cars made in association with Standard Motor Co, many bearing Jaguar as a model name. The company's name was changed from S. S. Cars to Jaguar Cars in 1945. A merger with the British Motor Corporation followed in 1966,[3] the resulting enlarged company now being renamed as British Motor Holdings (BMH), which in 1968 merged with Leyland Motor Corporation and became British Leyland, itself to be nationalised in 1975."

p23 = "Jaguar was spun off from British Leyland and was listed on the London Stock Exchange in 1984, becoming a constituent of the FTSE 100 Index until it was acquired by Ford in 1990.[4] Jaguar has, in recent years, manufactured cars for the British Prime Minister, the most recent delivery being an XJ in May 2010.[5] The company also holds royal warrants from Queen Elizabeth II and Prince Charles.[6]"

p24 = "Ford owned Jaguar Cars, also buying Land Rover in 2000, until 2008 when it sold both to Tata Motors. Tata created Jaguar Land Rover as a subsidiary holding company. At operating company level, in 2013 Jaguar Cars was merged with Land Rover to form Jaguar Land Rover Limited as the single design, manufacture, sales company and brand owner for both Jaguar and Land Rover vehicles."

p25 = "Since the Ford ownership era, Jaguar and Land Rover have used joint design facilities in engineering centres at Whitley in Coventry and Gaydon in Warwickshire and Jaguar cars have been assembled in plants at Castle Bromwich and Solihull."

p26 = "On 15 February 2021 Jaguar Land Rover announced that all cars made under the Jaguar brand will be fully electric by 2025.[7]"

#Our sentences we like to encode
sentences_2 = [p21, p22, p23, p24, p25, p26]
```

```python
from sentence_transformers import SentenceTransformer

import numpy as np
from numpy import linalg as LA
```

```python
model = SentenceTransformer('all-mpnet-base-v2')
```

```python
#Sentences are encoded by calling model.encode()
embeddings_1 = model.encode(sentences_1)
embeddings_2 = model.encode(sentences_2)
```

## Sum of vectors

```python
type(embeddings_1[0])
```

```python
embeddings_1[0].shape
```

```python
LA.norm(embeddings_1[0])
```

```python
for inx_1 in range(len(embeddings_1)):
  for inx_2 in range(len(embeddings_2)):
    vec_1 = embeddings_1[inx_1]
    vec_2 = embeddings_2[inx_2]
    sim = np.dot(vec_1, vec_2)
    print("sent {} from animal & sent {} from car sim: {}".format(inx_1, inx_2, sim))
```

```python
sum_vec_1 = embeddings_1[0]
for inx in range(1, len(embeddings_1)):
  sum_vec_1 += embeddings_1[inx]

sum_vec_2 = embeddings_2[0]
for inx in range(1, len(embeddings_2)):
  sum_vec_2 += embeddings_2[inx]


sum_vec_norm_1 = sum_vec_1 / np.linalg.norm(sum_vec_1)
sum_vec_norm_2 = sum_vec_2 / np.linalg.norm(sum_vec_2)
```

```python
LA.norm(sum_vec_norm_1)
```

```python
LA.norm(sum_vec_norm_2)
```

```python
tot_sim = np.dot(sum_vec_norm_1, sum_vec_norm_2)

print("dot product of sum of vecors: {}".format(tot_sim))
```
